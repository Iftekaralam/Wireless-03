```python

import numpy as np
import matplotlib.pyplot as plt

# Given values
C = 420
R = 1.0
N = np.array([3, 4, 7, 9, 12])

# Calculate reuse ratio
reuse_ratio = np.sqrt(3 * N)

# Calculate reuse distance
D = R * reuse_ratio

# Calculate channels per cell
channels_per_cell = C / N

# Display numerical results
print("CELLULAR FREQUENCY REUSE RESULTS")
print("---------------------------------------------")
print("N     D/R       D (km)       Channels/Cell")
print("---------------------------------------------")

for n, ratio, distance, channels in zip(
        N, reuse_ratio, D, channels_per_cell):

    print(f"{n:<5} {ratio:<10.2f} {distance:<13.2f} {channels:.1f}")

# Plot 1: Reuse Distance vs Cluster Size
plt.figure(figsize=(8, 5))
plt.plot(N, D, 'bo-', linewidth=2, markersize=8)
plt.xlabel("Cluster Size (N)")
plt.ylabel("Reuse Distance D (km)")
plt.title("Reuse Distance vs Cluster Size")
plt.grid(True)
plt.xticks(N)
plt.show()

# Plot 2: Channels per Cell vs Cluster Size
plt.figure(figsize=(8, 5))
plt.plot(N, channels_per_cell, 'ro-', linewidth=2, markersize=8)
plt.xlabel("Cluster Size (N)")
plt.ylabel("Channels per Cell")
plt.title("Channels per Cell vs Cluster Size")
plt.grid(True)
plt.xticks(N)
plt.show()

# Design Challenge
minimum_distance = 4.0

print("\nDESIGN CHALLENGE")
print("---------------------------------------------")

for n, distance, channels in zip(N, D, channels_per_cell):
    if distance >= minimum_distance:
        print("Smallest suitable cluster size:", n)
        print("Reuse distance:", round(distance, 2), "km")
        print("Channels per cell:", round(channels, 2))
        break
```

```python

import numpy as np

C = 420
R = 1.0
N = np.array([3, 4, 7, 9, 12])
minimum_distance = 4.0

D = R * np.sqrt(3 * N)
channels_per_cell = C / N

print("DESIGN CHALLENGE RESULTS")
print("-----------------------------------------")
print("N     D (km)     Channels per Cell")
print("-----------------------------------------")

for n, distance, channels in zip(N, D, channels_per_cell):
    print(f"{n:<5} {distance:<11.2f} {channels:.2f}")

for n, distance, channels in zip(N, D, channels_per_cell):
    if distance >= minimum_distance:
        print("\nSmallest feasible cluster size:", n)
        print("Reuse distance:", round(distance, 2), "km")
        print("Channels per cell:", round(channels, 2))
        break
```
