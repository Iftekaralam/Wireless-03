```python

import numpy as np
import matplotlib.pyplot as plt

c = 3e8
f = 900e6
Pt = 1.0
Gt = Gr = 1.0
ht = 20.0
hr = 2.0

d = np.linspace(10, 5000, 1000)
lam = c / f

Pr_fs = Pt * Gt * Gr * (lam / (4 * np.pi * d))**2
Pr_2ray = Pt * Gt * Gr * (ht**2 * hr**2) / (d**4)

d_break = 4 * np.pi * ht * hr / lam

print("Wavelength =", lam, "m")
print("Breakpoint distance =", d_break, "m")

plt.figure(figsize=(8, 5))
plt.semilogx(d, 10 * np.log10(Pr_fs) + 30, label="Free-space")
plt.semilogx(d, 10 * np.log10(Pr_2ray) + 30, label="Two-ray asymptotic")
plt.xlabel("Distance (m)")
plt.ylabel("Received Power (dBm)")
plt.title("Free-space vs Two-ray Propagation Model")
plt.grid(True)
plt.legend()
plt.show()

heights = [10, 20, 50]

plt.figure(figsize=(8, 5))

for h in heights:
    d_break_height = 4 * np.pi * h * hr / lam
    print("Transmitter Height =", h, "m, Breakpoint Distance =", d_break_height, "m")
    Pr = Pt * Gt * Gr * (h**2 * hr**2) / (d**4)
    plt.semilogx(d, 10 * np.log10(Pr) + 30, label=f"Height = {h} m")

plt.xlabel("Distance (m)")
plt.ylabel("Received Power (dBm)")
plt.title("Effect of Transmitter Height on Two-ray Propagation")
plt.grid(True)
plt.legend()
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

c = 3e8
f = 900e6
hr = 2.0
heights = np.array([10, 20, 50])

lam = c / f

breakpoint_distances = (4 * np.pi * heights * hr) / lam

print("BREAKPOINT DISTANCE COMPARISON")
print("---------------------------------------")
print("Height (m)     Breakpoint Distance (m)")
print("---------------------------------------")

for h, db in zip(heights, breakpoint_distances):
    print(f"{h:<15} {db:.2f}")

plt.figure(figsize=(8, 5))
plt.plot(heights, breakpoint_distances, 'bo-', linewidth=2, markersize=8)
plt.xlabel("Transmitter Height (m)")
plt.ylabel("Breakpoint Distance (m)")
plt.title("Transmitter Height vs Breakpoint Distance")
plt.grid(True)
plt.xticks(heights)
plt.show()

plt.figure(figsize=(8, 5))
plt.bar(heights.astype(str), breakpoint_distances)
plt.xlabel("Transmitter Height (m)")
plt.ylabel("Breakpoint Distance (m)")
plt.title("Comparison of Breakpoint Distances")
plt.grid(axis='y')
plt.show()

```
